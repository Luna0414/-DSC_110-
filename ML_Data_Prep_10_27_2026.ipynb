```python
import pandas as pd
#just selecting 3 specific things for the libary instead of using the whole libary to save storage
#make sure standard and scaler have captial S
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/osoliman/DSC110/refs/heads/main/Datasets/nhanes_final.csv')
```

```python
df.head()
```

Step 1: filtering

```python
df = df[df['age'] >=18]
#add space between to make it each step readable
```

```python
#getting minimum
df['age'].min()
```

Step 2: drop colums/fields

> Drop irrelevant fields e.g., id and fields that are not avabblie at the time of predicting hbalc



```python
#dropping any fields that are not relevant
# [] for the list
#if you run too many times it can give you an error
df = df.drop(columns=['id', 'hba1c', 'fasting_glucose'], errors='ignore')
```

```python
#check with df.head() if you get error to check if data was removed
# 3 and 4 where removed
df.head()
```

Step 3: mapping labels

```python
#creating dictionaries
gender_map = {1.0: 'male', 2.0: 'female'}
df['gender'] = df['gender'].map(gender_map)
```

```python
df['gender'].value_counts()
```

```python
#education
education_map = {
    1.0: 'less than 9th',
    2.0: '9th-11th grade',
    3.0: 'high school',
    4.0: 'some college',
    5.0: 'college grad',
    9.0: 'do not know'
}
```

```python
df['education'] = df['education'].map(education_map)
```

```python
df['education'].value_counts()
```

```python
#race_ethnicity mapping
#dictionary = {}
race_map = {
    1.0: 'Mexican American',
    2.0: 'Other Hispanic',
    3.0: 'White',
    4.0: 'Black',
    6.0: 'Asian',
    7.0: 'Other'
}
df['race_ethnicity'] = df['race_ethnicity'].map(race_map)
df['race_ethnicity'].value_counts()
```

Step 4: missing values

```python
#checking missing values
df.isna().sum()
```

```python
#a loop used instead of repeating this; df['bmi'] = df['bmi'].fillina
for col in ['bmi', 'height_cm', 'systolic_bp']:
  df[col] =df[col].fillna(df[col].mean())
```

```python
df.isna().sum()
```

```python
#removing rows where education is missing
# dropna =drops fields of values
df = df.dropna(subset=['education'])
```

```python
#now theres no missing values
df.isna().sum()
```

Step 5: define the target

```python
# step 1: include only rows that have diabetes yes or no
#removing raws where diabetes is borderlineg or missing
df = df[df['has_diabetes'].isin([1.0,2.0])]
```

```python
df['has_diabetes'].value_counts()
#2.0= no diabetes
#1.0= had diabetes
```

```python
def bp_check(systolic_bp):
  if systolic_bp > 120:
    bp_result = "high"
  else:
    bp_result = "normal"
  return bp_result
```

```python
bp_check(13)
```

```python
bp_check(120)
```

```python
bp_check(150)
```

```python
def BMI_check(bml_value):
  if BMI_value < 18:
     bml_result= "underweight"
  #more than one condition use elif
  elif BMI_value <25:
    BMI_result= "normal"
  elif BMI_value >30:
   BMI_result ="overweight"
  else:
    BMI_result = "obese"
  return BMI_result
```

```python
def calculate_BMI(weight,height):
  BMI = weight/(height **2)
  return BMI
```

```python
calculate_BMI(1.2,57)
```

```python
#define what you added to get the exact same answer
calculate_BMI(height=1.2, weight=57)
```
